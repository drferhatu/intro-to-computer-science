## 2 · Types and printf

Change the values, run again. What happens if you print an `int` with `%f`?

```python
%%writefile types.c
#include <stdio.h>

int main(void)
{
    int age = 19;
    float gpa = 3.75;
    char initial = 'F';
    printf("age %i, gpa %.2f, initial %c\n", age, gpa, initial);
    printf("an int takes %zu bytes, a float %zu, a char %zu, a double %zu\n",
           sizeof(int), sizeof(float), sizeof(char), sizeof(double));
}
```

```python
!clang -o types types.c && ./types
```

## 3 · A byte as bits (the Part D program, explained)

`(n >> i) & 1` shifts the number right by `i` places and keeps only the last bit. Watch it work for one number:

```python
%%writefile bits.c
#include <stdio.h>

int main(void)
{
    int n = 73;
    for (int i = 7; i >= 0; i--)
    {
        int bit = (n >> i) & 1;
        printf("i=%i  n>>i=%3i  bit=%i\n", i, n >> i, bit);
    }
}
```

```python
!clang -o bits bits.c && ./bits
```

## 4 · Integer overflow, on purpose

An `int` has 32 bits. What happens one past the largest value? (We meet this properly in week 3.)

```python
%%writefile overflow.c
#include <stdio.h>
#include <limits.h>

int main(void)
{
    int big = INT_MAX;
    printf("largest int : %i\n", big);
    big = big + 1;
    printf("plus one    : %i\n", big);
    unsigned char byte = 255;
    byte = byte + 1;
    printf("255 + 1 in one byte: %i\n", byte);
}
```

```python
!clang -Wno-integer-overflow -o overflow overflow.c && ./overflow
```

The bits wrap around, like an odometer. This exact behavior grounded the Boeing 787 (week 3's wow moment).
